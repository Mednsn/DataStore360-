```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from ydata_profiling import ProfileReport


df = pd.read_csv("../data/raw/dataset.csv")

# print(df.head())
# print(df.shape)
# print(df.columns)
# print(df.dtypes)
df["Order ID"] = df["Order ID"].astype("string")
df["Ship Date"] = pd.to_datetime(df["Ship Date"])
df["Ship Mode"] = df["Ship Mode"].astype("string")
df["Customer ID"] = df["Customer ID"].astype("string")
df["Customer Name"] = df["Customer Name"].astype("string")
df["Segment"] = df["Segment"].astype("string")
df["Country"] = df["Country"].astype("string")
df["City"] = df["City"].astype("string")
df["State"] = df["State"].astype("string")
df["Region"] = df["Region"].astype("string")
df["Product ID"] = df["Product ID"].astype("string")
df["Category"] = df["Category"].astype("string")
df["Sub-Category"] = df["Sub-Category"].astype("string")
df["Product Name"] = df["Product Name"].astype("string")
df["Sales"] = df["Sales"].astype("float")
df["Profit"] = df["Profit"].astype("float")

df["Order Date"] = pd.to_datetime(df["Order Date"], errors="coerce")

# print(df["Customer Name"].dtype)
# print("les valeur vide :",df.isna().sum())
# print("duplicate :",df.duplicated().sum())
# print(df.describe())





```

```python
profile = ProfileReport(df, title="Dataset Profiling Report")

profile.to_file("profiling_report.html")
```

```python
profit_by_category = df.groupby("Category")["Profit"].sum()

print("analyse profit par category : \n",profit_by_category)
```

```python
sale_by_region = df.groupby("Region")["Sales"].sum()
sales_by_region = sale_by_region.sort_values(ascending=False)

print("analyse de sale par region :\n",sales_by_region)  
```

```python
sales_by_segment = df.groupby("Segment")["Sales"].sum()

print("analyse de sale par segment :\n",sales_by_segment)

```

```python
# Observation de figure est ce que sales lier avec profit et est ce que ils sont augmente en meme temps :

plt.figure()
plt.scatter(df["Sales"],df["Profit"])
plt.title("Sales vs Profits")
plt.xlabel("Sales")
plt.xlim(0, 10000) 
plt.ylabel("Profits")
plt.ylim(-10000, 10000) 
plt.show()
```

```python
# pour savoir la distribution de sales (augmantation , des outliyers et grand ou petit):

plt.figure()
plt.hist(df["Sales"], bins=30)
plt.title("Distribution of Sales")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.show()
```
