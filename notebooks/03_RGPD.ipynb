```python
import pandas as pd
import hashlib

file_path = "../data/processed/dataset_cleaned.csv"

df_cleaned = pd.read_csv(file_path)

print(df_cleaned.shape)
df_cleaned.columns


```

```python
print("Nombre de valeurs manquantes :", df_cleaned["Customer Name"].isna().sum())
print("Nombre de clients uniques :", df_cleaned["Customer Name"].nunique())

df_cleaned[["Customer ID", "Customer Name"]].head(10)
```

```python
def pseudonymize(value):
    if pd.isna(value):
        return value
    
    return hashlib.sha256(str(value).encode()).hexdigest()  
```

```python
test_name = "Claire Gute"

print("Nom original :", test_name)
print("Nom pseudonymise :", pseudonymize(test_name))
```

```python
df_cleaned["Customer Name"] = df_cleaned["Customer Name"].apply(pseudonymize)
```
